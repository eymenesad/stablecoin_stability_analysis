# Stablecoin Stability Analysis - Modeling Results

This notebook analyzes the results of our machine learning models: Clustering, Classification, Anomaly Detection, and Regression.

```python
import pandas as pd
import plotly.express as px

RESULTS_PATH = '../reports/metrics/modeling_results.csv'
df = pd.read_csv(RESULTS_PATH)
df.head()
```

## 1. Clustering Analysis (K-Means)
Visualizing the clusters found by K-Means. We use `volatility_24h` and `btc_close` to see if clusters separate market conditions.

```python
fig = px.scatter(df, x='volatility_24h', y='btc_close', color='cluster', 
                 title='K-Means Clusters (Volatility vs BTC Price)',
                 hover_data=['symbol', 'timestamp'])
fig.show()
```

## 2. Anomaly Detection (Isolation Forest)
Visualizing the anomalies detected by the model. Anomalies (value -1) are potential depeg events or extreme market conditions.

```python
anomalies = df[df['anomaly'] == -1]
fig = px.scatter(df, x='timestamp', y='close', color='anomaly', 
                 title='Isolation Forest Anomalies (Price History)',
                 color_continuous_scale=['red', 'blue'])
fig.show()
```

## 3. Classification Results
Analyzing the distribution of predicted stability labels.

```python
fig = px.histogram(df, x='stability_label', color='symbol', title='Stability Label Distribution')
fig.show()
```
