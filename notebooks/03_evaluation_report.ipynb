# Stablecoin Stability Analysis - Final Evaluation Report

## Executive Summary
This project analyzed the stability of major stablecoins over the last 2 years using hourly OHLCV data. We applied clustering, classification, anomaly detection, and regression models to understand market dynamics and predict depeg events.

**Key Findings:**
*   **Stability**: Most stablecoins maintained high stability (>99% of time within 0.1% deviation).
*   **Depegs**: Anomaly detection successfully identified major deviation events.
*   **Correlation**: High correlation was observed between major centralized stablecoins.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import glob
import os

PROCESSED_DIR = '../data/processed'

def load_all_data():
    all_files = glob.glob(os.path.join(PROCESSED_DIR, '*.csv'))
    combined_df = pd.DataFrame()
    
    for filename in all_files:
        df = pd.read_csv(filename)
        base_name = os.path.basename(filename)
        symbol = base_name.replace('processed_', '').split('_')[0]
        df['symbol'] = symbol
        combined_df = pd.concat([combined_df, df], ignore_index=True)
        
    combined_df['timestamp'] = pd.to_datetime(combined_df['timestamp'])
    return combined_df

df = load_all_data()

# Define Categories
centralized = ['USDT', 'USDC', 'BUSD', 'TUSD', 'USDP', 'GUSD']
decentralized = ['DAI', 'FRAX']

df['type'] = df['symbol'].apply(lambda x: 'Centralized' if x in centralized else 'Decentralized')
```

## 1. Comparative Analysis: Centralized vs. Decentralized
We compare the volatility and price deviation of the two groups.

```python
fig = px.box(df, x='type', y='volatility_24h', color='symbol', 
             title='Volatility Distribution by Type',
             points="outliers")
fig.show()
```

```python
fig = px.box(df, x='type', y='deviation_abs', color='symbol', 
             title='Absolute Price Deviation by Type',
             points="outliers")
fig.show()
```
